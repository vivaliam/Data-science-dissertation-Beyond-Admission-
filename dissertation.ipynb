```python
import pandas as pd

adm = pd.read_csv(r"C:\Users\hp\Downloads\international_postgraduate_admissions_2015_2025.csv")
emp = pd.read_csv(r"C:\Users\hp\Downloads\international_pg_employment_2014_2021.csv")

# Standardise year formats: "2015/16" and "2014/2015" -> "2015/16" style
def clean_year(y):
    y = str(y)
    start = y[:4]
    return f"{start}/{str(int(start)+1)[2:]}"

adm['year'] = adm['Academic Year'].apply(clean_year)
emp['year'] = emp['academic_year'].apply(clean_year)

merged = pd.merge(adm[['year', 'Number']], 
                  emp[['year', 'total_grads', 'emp_rate_weighted']], 
                  on='year', how='outer').sort_values('year').reset_index(drop=True)

merged.columns = ['year', 'admissions', 'graduates', 'emp_rate']
print(merged)

merged.to_csv(r"C:\Users\hp\Downloads\merged_analysis_dataset.csv", index=False)
print("\nSaved successfully.")
```

```python
import pandas as pd
import numpy as np

merged = pd.read_csv(r"C:\Users\hp\Downloads\merged_analysis_dataset.csv")
adm = merged.dropna(subset=['admissions']).copy().reset_index(drop=True)

adm['t'] = range(len(adm))
train, test = adm.iloc[:8], adm.iloc[8:]
print("Train years:", train['year'].tolist())
print("Test years: ", test['year'].tolist())
```

```python
from statsmodels.tsa.arima.model import ARIMA
from statsmodels.tsa.holtwinters import ExponentialSmoothing
from prophet import Prophet

results = {}

# ARIMA(1,1,1) - modest order given short series
arima_fit = ARIMA(train['admissions'], order=(1,1,1)).fit()
results['ARIMA'] = arima_fit.forecast(steps=2).values

# Exponential Smoothing (Holt's linear trend)
es_fit = ExponentialSmoothing(train['admissions'], trend='add').fit()
results['ExpSmoothing'] = es_fit.forecast(2).values

# Prophet - needs ds/y date format
ptrain = pd.DataFrame({
    'ds': pd.to_datetime([f"{y[:4]}-09-01" for y in train['year']]),
    'y': train['admissions'].values
})
m = Prophet(yearly_seasonality=False, weekly_seasonality=False, daily_seasonality=False)
m.fit(ptrain)
future = pd.DataFrame({'ds': pd.to_datetime([f"{y[:4]}-09-01" for y in test['year']])})
results['Prophet'] = m.predict(future)['yhat'].values

# Evaluate
actual = test['admissions'].values
print(f"\nActual test values: {actual}")
for name, preds in results.items():
    rmse = np.sqrt(np.mean((preds - actual)**2))
    mae = np.mean(np.abs(preds - actual))
    print(f"\n{name}: predictions = {preds.round(0)}")
    print(f"  RMSE = {rmse:,.0f}   MAE = {mae:,.0f}")
```

```python
full = adm.copy()

pfull = pd.DataFrame({
    'ds': pd.to_datetime([f"{y[:4]}-09-01" for y in full['year']]),
    'y': full['admissions'].values
})
m2 = Prophet(yearly_seasonality=False, weekly_seasonality=False, daily_seasonality=False)
m2.fit(pfull)
future_years = ['2025/26', '2026/27', '2027/28']
fut = pd.DataFrame({'ds': pd.to_datetime([f"{y[:4]}-09-01" for y in future_years])})
fc = m2.predict(fut)

print("Prophet forecasts (selected model), trained on full 2015/16-2024/25 series:")
for yr, row in zip(future_years, fc.itertuples()):
    print(f"  {yr}: {row.yhat:,.0f}   (80% interval: {row.yhat_lower:,.0f} to {row.yhat_upper:,.0f})")

# Comparison: ARIMA and ExpSmoothing on full series
arima_full = ARIMA(full['admissions'], order=(1,1,1)).fit().forecast(3).values
es_full = ExponentialSmoothing(full['admissions'], trend='add').fit().forecast(3).values
print("\nARIMA:        ", [f"{v:,.0f}" for v in arima_full])
print("ExpSmoothing: ", [f"{v:,.0f}" for v in es_full])
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ---- 1. Validation metrics table (Table 4.5) ----
actual = test['admissions'].values
val_rows = []
for name, preds in results.items():
    val_rows.append({
        'model': name,
        'pred_2023_24': round(preds[0]),
        'pred_2024_25': round(preds[1]),
        'RMSE': round(np.sqrt(np.mean((preds - actual)**2))),
        'MAE': round(np.mean(np.abs(preds - actual)))
    })
validation_table = pd.DataFrame(val_rows)
validation_table.to_csv(r"C:\Users\hp\Downloads\table_4_5_validation_metrics.csv", index=False)
print("Saved: table_4_5_validation_metrics.csv")
print(validation_table.to_string(index=False))

# ---- 2. Forecast table (Table 4.6) ----
forecast_table = pd.DataFrame({
    'year': ['2025/26', '2026/27', '2027/28'],
    'prophet': fc['yhat'].round(0).values,
    'prophet_lower_80': fc['yhat_lower'].round(0).values,
    'prophet_upper_80': fc['yhat_upper'].round(0).values,
    'arima': np.round(arima_full),
    'exp_smoothing': np.round(es_full)
})
forecast_table.to_csv(r"C:\Users\hp\Downloads\table_4_6_forecasts.csv", index=False)
print("\nSaved: table_4_6_forecasts.csv")
print(forecast_table.to_string(index=False))

# ---- 3. Forecast chart (Figure 4.3) ----
fig, ax = plt.subplots(figsize=(10, 6))
ax.plot(full['year'], full['admissions'], marker='o', color='#1f4e79',
        linewidth=2, label='Observed admissions')

fy = forecast_table['year'].tolist()
ax.plot(fy, forecast_table['prophet'], marker='o', color='#c0504d',
        linewidth=2, linestyle='--', label='Prophet forecast (selected)')
ax.fill_between(fy, forecast_table['prophet_lower_80'], forecast_table['prophet_upper_80'],
                color='#c0504d', alpha=0.15, label='Prophet 80% interval')
ax.plot(fy, forecast_table['arima'], marker='^', color='#4f6228',
        linewidth=1.5, linestyle=':', label='ARIMA(1,1,1)')
ax.plot(fy, forecast_table['exp_smoothing'], marker='s', color='#7f7f7f',
        linewidth=1.5, linestyle=':', label='Exponential Smoothing')

ax.axvline(x=9.5, color='gray', linestyle='-', linewidth=0.8, alpha=0.5)
ax.text(9.55, ax.get_ylim()[0] + 20000, 'Forecast →', fontsize=9, color='gray')
ax.set_xlabel('Academic year')
ax.set_ylabel('International postgraduate enrolments')
ax.set_title('Observed admissions (2015/16–2024/25) and model forecasts to 2027/28')
ax.tick_params(axis='x', rotation=45)
ax.legend(loc='upper left')
fig.tight_layout()
plt.savefig(r"C:\Users\hp\Downloads\figure_4_3_forecasts.png", dpi=300, bbox_inches='tight')
plt.show()
print("\nSaved: figure_4_3_forecasts.png")

# ---- 4. Regression scatter chart (Figure 4.4, optional for 4.4 section) ----
overlap = merged.dropna(subset=['admissions', 'emp_rate'])
fig2, ax2 = plt.subplots(figsize=(8, 5.5))
ax2.scatter(overlap['admissions'], overlap['emp_rate'], color='#1f4e79', s=70, zorder=3)
for _, r in overlap.iterrows():
    ax2.annotate(r['year'], (r['admissions'], r['emp_rate']),
                 textcoords="offset points", xytext=(8, 4), fontsize=9)
slope, intercept = 0.00001352, 10.692
xs = np.linspace(overlap['admissions'].min(), overlap['admissions'].max(), 100)
ax2.plot(xs, intercept + slope*xs, color='#c0504d', linestyle='--',
         label='Fitted line (R²=0.40, p=0.18)')
ax2.set_xlabel('International postgraduate admissions')
ax2.set_ylabel('Sustained employment rate at 1 YAG (%)')
ax2.set_title('Admissions vs employment rate, overlap window 2015/16–2020/21')
ax2.legend()
fig2.tight_layout()
plt.savefig(r"C:\Users\hp\Downloads\figure_4_4_regression_scatter.png", dpi=300, bbox_inches='tight')
plt.show()
print("Saved: figure_4_4_regression_scatter.png")
```
